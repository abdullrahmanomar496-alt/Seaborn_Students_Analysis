## 7. Student distribution by Study Time

```python
sns.countplot(data = df , x = "studytime" , color = "purple")
plt.title("Number of students by study time")
plt.xlabel("Study Time")
plt.ylabel("Number of students")
plt.show()
```

## 8. Box plot
Study time and Final grades

```python
sns.boxplot(data = df , x = "studytime" , y = 'G3' , color = "violet")
plt.title("Study Time Vs Final Math Grade ")
plt.xlabel("Study Time")
plt.ylabel("Final Math Grade")
plt.show()
```

## 9. Scatter Plot
Relationship Between Absences and Final Grades

```python
sns.scatterplot(data = df , x = "absences" , y = "G3" , color="black")
plt.title("Absences VS Final Math Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Math Grade")
plt.show()
```

## 10. Correlation Heatmap
correlation between numerical variables

```python
correlation = df.corr(numeric_only = True)   # Show the Numeric correlation
correlation
```

```python
plt.figure(figsize=(12 , 8))  # Heatmap Graph
sns.heatmap(correlation , annot = True , cmap = "coolwarm" , fmt = ".2f")
plt.title("correlation Heatmap of Student Performance")
plt.show()
```

## 11. Violin Plot
Distribution of final Grades by Study Time

```python
sns.violinplot(data = df , x = "studytime" , y = "G3" ,color = "silver")
plt.title("Distribution of Math Grade By study time")
plt.xlabel("Study Time")
plt.ylabel("Final Math Grade")
plt.show()
```

## 12. Comparing Grades by study Time and Gander

```python
sns.boxplot(data = df , x = "studytime" , y = "G3" , hue = "sex") # hue is the column we compare by it
plt.title("Final math grade by study time and gendr")
plt.xlabel("Study Time")
plt.ylabel("Final Math Grade")
plt.show()
```

## 13. Pair Plot
Exploring Relationships Between Variables

```python
sns.pairplot(df[["studytime" , "absences" , "G1" , "G2" , "G3"]], diag_kind = "hist")
plt.show()
```
